# KERASCAN Phase 1 image engine

Research-only execution notebook. The supplied images are inference-only; no cell fits on them or assigns a clinical diagnosis.

```python
from pathlib import Path
import sys
ROOT = Path.cwd()
if not (ROOT / 'src').exists(): ROOT = ROOT.parent
sys.path.insert(0, str(ROOT / 'src'))
from kerascan.inference import KerascanEngine
from kerascan.image_io import read_image
engine = KerascanEngine()
samples = [ROOT.parent / 'sample_images' / 'aleft.png', ROOT.parent / 'sample_images' / 'aright.png']
[(p.name, read_image(p).shape) for p in samples]
```

```python
# This fits only deterministic synthetic development proxies.
metadata = engine.fit_synthetic_baseline()
metadata
```

```python
summaries = {}
for image_path in samples:
    result = engine.analyze(image_path, ROOT / 'outputs' / image_path.stem)
    summaries[image_path.name] = {
        'roi': result['roi'],
        'quality': result['quality'],
        'screening_result': result['screening_result'],
        'experimental_status': result['experimental_status'],
    }
summaries
```

`NORMAL-LIKE` and `SUSPICIOUS` are experimental prototype-screening outputs only. `UNGRADABLE` means recapture required and stops before classification. Arc-Step physical topography is deliberately out of scope until calibrated KERASCAN camera parameters, working distance, and Placido geometry are available.
